```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
import pickle
```

```python
import tensorflow as tf
print(tf.__version__)
```

## Load The Dataset

```python
## Load The dataset
df=pd.read_csv("Churn_Modelling.csv")
df.head()
```

### Preprocess the data
### Drop irrelevant columns

```python
df=df.drop(['RowNumber','CustomerId','Surname'],axis=1)
df
```

## Encode one categorical variables

```python
label_encoder_gender=LabelEncoder()
df['Gender']=label_encoder_gender.fit_transform(df['Gender'])
df
```

### Onehot encoder 'Geograpy


```python
from sklearn.preprocessing import OneHotEncoder
onehotencoder=OneHotEncoder()
geo_encoder=onehotencoder.fit_transform(df[['Geography']])
geo_encoder
```

```python
geo_encoder.toarray()
```

```python
onehotencoder.get_feature_names_out(['Geography'])
```

```python
geo_encoder_df=pd.DataFrame(geo_encoder.toarray(),columns=onehotencoder.get_feature_names_out(['Geography']))
```

```python
geo_encoder_df
```

### Combine the one hot encoder colimns with the original data

```python
df=pd.concat([df.drop('Geography', axis=1),geo_encoder_df], axis=1)
```

```python
df.head()
```

### Save the encoder and scaler

```python
with open('label_encoder_gender.pkl','wb') as file:
    pickle.dump(label_encoder_gender,file)

with open('onehotencoder.pkl','wb') as file:
    pickle.dump(onehotencoder,file)
    
```

### Divide the dataset into indepent and dependent features

```python
X=df.drop('Exited', axis=1)
y=df['Exited']

## Split the data with training and testing sets
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.2)
```

```python
## Scaling the features
scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
X_train
```

```python
with open('scaler.pkl','wb') as file:
    pickle.dump(scaler,file)
```

```python
df
```

## ANN Implementation

```python
import tensorflow as ts
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime
```

## Build Our ANN

```python

# Define the model
model = Sequential([
    Dense(64, activation='relu', input_shape=(X_train.shape[1],)),  # 1st hidden layer connected with input layer
    Dense(32, activation='relu'),  # Hidden layer 2
    Dense(1, activation='sigmoid')  # Output layer
])
```

```python
model.summary()
```

## Compile the model

```python
import tensorflow
opt=tensorflow.keras.optimizers.Adam(learning_rate=0.01)
loss=tensorflow.keras.losses.BinaryCrossentropy()
loss
```

```python
model.compile(optimizer=opt,loss="binary_crossentropy",metrics=['accuracy'])
```

### Setup the Tensorbroad

```python
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard

log_dir="logs/fit" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorflow_callback=TensorBoard(log_dir=log_dir,histogram_freq=1)
```

### Set up Early Stopping

```python
early_stopping=EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)
```

### Training The Model

```python
history=model.fit(X_train,y_train,validation_data=(X_test,y_test),epochs=100,
                  callbacks=[tensorflow_callback,early_stopping]
                  )
```

```python
model.save('model.h5')
```
