<h1>LSTM কী এবং কেন এটি ব্যবহার করা হয়?</h1>

<h2>১. LSTM কী?</h2>
<p>
LSTM এর পূর্ণরূপ হলো <b>Long Short-Term Memory</b>। এটি একটি বিশেষ ধরনের <b>RNN (Recurrent Neural Network)</b> যা দীর্ঘ সময়ের জন্য তথ্য মনে রাখতে পারে।
</p>

<h2>২. RNN কী?</h2>
<p>
RNN এমন একটি নিউরাল নেটওয়ার্ক যা ধারাবাহিক (sequence) ডেটা যেমন ভাষা, সময়-সারি, অডিও ইত্যাদির জন্য ব্যবহৃত হয়। এটি পূর্ববর্তী ইনপুটের তথ্য পরবর্তী ইনপুটে ব্যবহার করতে পারে।
</p>

<h3>❌ RNN-এর সমস্যাসমূহ:</h3>
<ul>
  <li>দীর্ঘমেয়াদী তথ্য মনে রাখতে পারে না</li>
  <li>ভ্যানিশিং গ্র্যাডিয়েন্ট সমস্যা</li>
  <li>পুরানো তথ্য ভুলে যায়</li>
</ul>

<h2>৩. LSTM কেন ব্যবহার করবো?</h2>
<p>
LSTM মূলত RNN-এর এই সমস্যা গুলো সমাধান করে। এটি তিনটি 'গেট'-এর মাধ্যমে কাজ করে:
</p>

<ul>
  <li><b>Forget Gate:</b> কোন তথ্য ভুলে যাবে তা নির্ধারণ করে</li>
  <li><b>Input Gate:</b> কোন নতুন তথ্য মেমোরিতে সংরক্ষণ করা হবে</li>
  <li><b>Output Gate:</b> কোন তথ্য বাইরে পাঠানো হবে তা নির্ধারণ করে</li>
</ul>

<h2>৪. বাস্তব জীবনের উপমা</h2>
<p>
ধরো তুমি একটা গল্প পড়ছো। তুমি প্রতিটি শব্দ মনে রাখো না, কিন্তু গুরুত্বপূর্ণ বিষয়গুলো মনে রাখো — যেমন গল্পের চরিত্র বা প্লট। LSTM ঠিক এভাবেই গুরুত্বপূর্ণ তথ্য মনে রাখে এবং অপ্রয়োজনীয় তথ্য ভুলে যায়।
</p>

<h2>৬. উদাহরণ (Keras কোড):</h2>
<pre>
from keras.models import Sequential
from keras.layers import Embedding, LSTM, Dense

model = Sequential()
model.add(Embedding(input_dim=10000, output_dim=128, input_length=100))
model.add(LSTM(units=64))
model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
</pre>
